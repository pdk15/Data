```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
df=pd.read_excel('EDA_Source/flight_price.xlsx')
df.head()
```

```python
df.info()
```

```python
## now doing feature engineering my converting data object into  integer format
## first we split all the dates in date , month , year

 # we got by this 1st index of data_of_journy string i.e date
df['Date']=df['Date_of_Journey'].str.split('/').str[0]

 # we got by this 1st index of data_of_journy string i.e Month
df['Month']=df['Date_of_Journey'].str.split('/').str[1]


```

```python

 # we got by this 2th index of data_of_journy string i.e Year
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.info()
```

```python
df.head()
```

```python
## Now we convert that all dates objects into integer

df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)

df.info()
```

```python
## Now we drop Date_of_journy column 
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_Hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_Min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.drop('Arrival_Time',axis=1, inplace=True)
```

```python
df.info()
```

```python
df['Arrival_Hour']=df['Arrival_Hour'].astype(int)
df['Arrival_Min']=df['Arrival_Min'].astype(int)
```

```python
df.info()
```

```python
df.head()
```

```python
## now convert departure time in int
df['Departure_hour']=df['Dep_Time'].str.split(':').str[0]
df['Departure_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df['Departure_hour']=df['Departure_hour'].astype(int)
df['Departure_min']=df['Departure_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Duration_Hour']=df['Duration'].str.split('h').str[0]
df['Duration_Min']=df['Duration'].str.split(' ').str[1].str.replace('m','',regex=False)
```

```python
df.info()
```

```python
df['Duration_Hour'] = df['Duration'].str.extract(r'(\d+)h').fillna(0).astype(int) #r'(\d+)h' extracts the digits before h.
df['Duration_Min'] = df['Duration'].str.extract(r'(\d+)m').fillna(0).astype(int)  #r'(\d+)m' extracts the digits before m.
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.head()
```
